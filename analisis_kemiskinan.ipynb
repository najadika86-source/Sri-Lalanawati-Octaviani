```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data_kemiskinan.csv", sep=";")

df.head()
```

```python
df.columns
```

```python
top5 = df.sort_values("2021", ascending=False).head(5)

top5[["Kab/Kota", "2021"]]
```

```python
plt.figure(figsize=(10,6))

bars = plt.bar(top5["Kab/Kota"], top5["2021"])

plt.xlabel("Kabupaten/Kota")
plt.ylabel("Persentase Penduduk Miskin (%)")
plt.title("5 Kabupaten/Kota dengan Tingkat Kemiskinan Tertinggi Tahun 2021")

plt.xticks(rotation=45)

for bar in bars:
    nilai = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        nilai,
        f"{nilai:.2f}%",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```
